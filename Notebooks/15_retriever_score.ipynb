```python
from dotenv import load_dotenv, find_dotenv
import os
from llama_index.core import VectorStoreIndex, SimpleDirectoryReader
from llama_index.core.node_parser import SentenceSplitter


# Load environment variables
_ = load_dotenv(find_dotenv())


def load_doc(file):
    documents = SimpleDirectoryReader(input_files=[file]).load_data()
    node_parser = SentenceSplitter(chunk_size=512)
    nodes = node_parser.get_nodes_from_documents(documents)
    
    return nodes

def nodes(nodes):
    for idx, node in enumerate(nodes):
        node.id_ = f"node_{idx}"
    return nodes

def vectorstore(nodes):
    index = VectorStoreIndex(nodes)
    return index
def index_retriever(index):
    retriever = index.as_retriever(similarity_top_k=5)
    return retriever
```

```python
document = load_doc("../data/Pmg_lds.md")
nodes = nodes(document)
vector = vectorstore(nodes)
retriever = index_retriever(vector)
```

```python
response = retriever.retrieve("describe the book of mormon")
new_list = [i for i in response]

len(new_list)
```
